```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, roc_curve

#==========================
# Prepare Data
#==========================

X = df.drop('readmitted', axis=1)
y = df['readmitted']

X = pd.get_dummies(
    X,
    columns=['gender',
             'primary_diagnosis',
             'discharge_to'],
    drop_first=True
)

#==========================
# Train Test Split
#==========================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

#==========================
# Logistic Regression
#==========================

log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model.fit(X_train, y_train)

#==========================
# Random Forest
#==========================

rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=5,
    random_state=42
)

rf.fit(X_train, y_train)

#==========================
# Prediction Probabilities
#==========================

log_prob = log_model.predict_proba(X_test)[:,1]

rf_prob = rf.predict_proba(X_test)[:,1]

#==========================
# ROC AUC
#==========================

auc_log = roc_auc_score(y_test, log_prob)
auc_rf = roc_auc_score(y_test, rf_prob)

print("Logistic Regression AUC =", auc_log)
print("Random Forest AUC =", auc_rf)

#==========================
# ROC Curve
#==========================

fpr_log, tpr_log, _ = roc_curve(y_test, log_prob)

fpr_rf, tpr_rf, _ = roc_curve(y_test, rf_prob)

plt.figure(figsize=(7,6))

plt.plot(
    fpr_log,
    tpr_log,
    label=f'Logistic Regression (AUC={auc_log:.3f})'
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f'Random Forest (AUC={auc_rf:.3f})'
)

plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()

plt.show()

#==========================
# Cross Validation
#==========================

cv_log = cross_val_score(
    log_model,
    X,
    y,
    cv=5,
    scoring='roc_auc'
)

cv_rf = cross_val_score(
    rf,
    X,
    y,
    cv=5,
    scoring='roc_auc'
)

print("\nCross Validation Results")
print("------------------------")
print("Logistic Regression Mean AUC :", cv_log.mean())
print("Random Forest Mean AUC       :", cv_rf.mean())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import roc_auc_score

df = pd.read_csv('train_df.csv')
numeric_cols = ['age', 'num_procedures', 'days_in_hospital', 'comorbidity_score']

X = pd.get_dummies(
    df[numeric_cols + ['gender', 'primary_diagnosis', 'discharge_to']],
    columns=['gender', 'primary_diagnosis', 'discharge_to'],
    drop_first=True
).astype(float)
y = df['readmitted'].values

X_std = StandardScaler().fit_transform(X.values)


rf = RandomForestClassifier(n_estimators=300, max_depth=5, min_samples_leaf=20, random_state=42, n_jobs=-1)
rf.fit(X, y)
rf_full_auc = roc_auc_score(y, rf.predict_proba(X)[:, 1])

logit = LogisticRegression(max_iter=1000)
logit.fit(X_std, y)
logit_full_auc = roc_auc_score(y, logit.predict_proba(X_std)[:, 1])

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rf_cv_auc = cross_val_score(rf, X.values, y, cv=skf, scoring='roc_auc')
logit_cv_auc = cross_val_score(logit, X_std, y, cv=skf, scoring='roc_auc')

print("Random Forest    - Full Data AUC:", round(rf_full_auc, 4), "| CV AUC (mean):", round(rf_cv_auc.mean(), 4))
print("Logistic Regr.   - Full Data AUC:", round(logit_full_auc, 4), "| CV AUC (mean):", round(logit_cv_auc.mean(), 4))

labels = ['Logistic\n(Full Data)', 'Logistic\n(5-Fold CV)',
          'Random Forest\n(Full Data)', 'Random Forest\n(5-Fold CV)']
values = [logit_full_auc, logit_cv_auc.mean(), rf_full_auc, rf_cv_auc.mean()]
colors = ['#2E5A88', '#2E5A88', '#C1443C', '#C1443C']
alphas = [1, 0.5, 1, 0.5]

fig, ax = plt.subplots(figsize=(7.5, 5.5))
bars = ax.bar(labels, values, color=colors)

for bar, val, a in zip(bars, values, alphas):
    bar.set_alpha(a)
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.01, f'{val:.3f}',
            ha='center', fontweight='bold', fontsize=9)

ax.axhline(y=0.5, color='gray', linestyle='--', linewidth=1)
ax.set_ylabel('ROC-AUC')
ax.set_title('Full-Data Evaluation vs. Cross-Validation (Overfitting Check)')
ax.set_ylim(0, 0.85)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA


df = pd.read_csv('train_df.csv')
numeric_cols = ['age', 'num_procedures', 'days_in_hospital', 'comorbidity_score']

X = df[numeric_cols].values


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


K_range = range(2, 9)
wcss = []
silhouette_scores = []

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    wcss.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))
    print(f"k={k}  |  WCSS={km.inertia_:.1f}  |  Silhouette={silhouette_score(X_scaled, labels):.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))


axes[0].plot(list(K_range), wcss, marker='o', color='#2E5A88')
axes[0].set_title('Elbow Method (WCSS)')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Within-Cluster Sum of Squares')


axes[1].plot(list(K_range), silhouette_scores, marker='o', color='#C1443C')
axes[1].set_title('Silhouette Score by k')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Average Silhouette Score')
axes[1].axhline(y=0.25, color='gray', linestyle='--', linewidth=1, label='Weak-structure threshold (0.25)')
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.show()
```

```python
k_final = 4
km_final = KMeans(n_clusters=k_final, random_state=42, n_init=10)
cluster_labels = km_final.fit_predict(X_scaled)

df['cluster'] = cluster_labels  

print("عدد المرضى في كل Cluster:")
print(df['cluster'].value_counts().sort_index())
```

```python
pca = PCA(n_components=2)
pcs = pca.fit_transform(X_scaled)

print("نسبة التباين اللي كل محور بيفسرها:", pca.explained_variance_ratio_)
print("المجموع:", pca.explained_variance_ratio_.sum())
```

```python
colors = ['#2E5A88', '#C1443C', '#4F9D69', '#E0A32E']

fig, ax = plt.subplots(figsize=(7.5, 6))
for c in range(k_final):
    mask = cluster_labels == c
    ax.scatter(pcs[mask, 0], pcs[mask, 1], s=8, alpha=0.5, color=colors[c], label=f'Cluster {c} (n={mask.sum()})')

ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)')
ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)')
ax.set_title(f'K-Means Clusters (k={k_final}) — PCA Projection')
ax.legend(markerscale=2, fontsize=9)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy import stats

numeric_cols = ['age', 'num_procedures', 'days_in_hospital', 'comorbidity_score']
Y = df[numeric_cols].values
groups = df['cluster'].values
k = 4
N, p = Y.shape   

y_bar = Y.mean(axis=0) 


B = np.zeros((p, p))
W = np.zeros((p, p))

for g in range(k):
    Yg = Y[groups == g]
    ng = len(Yg)
    yg_bar = Yg.mean(axis=0)
    
    diffB = (yg_bar - y_bar).reshape(-1, 1)
    B += ng * diffB @ diffB.T
    
    for row in Yg:
        d = (row - yg_bar).reshape(-1, 1)
        W += d @ d.T

wilks_lambda = np.linalg.det(W) / np.linalg.det(W + B)
print("Wilks' Lambda:", wilks_lambda)
```

```python
df_manova = p * (k - 1)  
chi2_manova = -(N - 1 - (p + k) / 2) * np.log(wilks_lambda)
p_manova = 1 - stats.chi2.cdf(chi2_manova, df_manova)

print("Chi-Square:", chi2_manova, "| df:", df_manova, "| p-value:", p_manova)
```
